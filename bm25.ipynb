```python
from rank_bm25 import BM25Okapi
from collections import Counter
import math
```

```python
# -----------------------------------
# STEP 1: CREATE 10 DOCUMENTS
# -----------------------------------

documents = [
    "Machine learning helps computers learn from data",
    "Deep learning uses neural networks for machine learning",
    "Natural language processing helps computers understand language",
    "Artificial intelligence uses machine learning and deep learning",
    "Neural networks are important for deep learning",
    "Python programming is useful for machine learning",
    "Language models use natural language processing",
    "Computer vision helps computers understand images",
    "Machine learning models use data for prediction",
    "Artificial intelligence improves computer vision and robotics"
]
```

```python
# -----------------------------------
# STEP 2: TOKENIZATION
# -----------------------------------

tokenized_docs = [
    doc.lower().split()
    for doc in documents
]

print("FIRST TOKENIZED DOCUMENT:")
print(tokenized_docs[0])
```

```python
# -----------------------------------
# STEP 3: VOCABULARY EXTRACTION
# -----------------------------------

vocabulary = sorted(
    set(word for doc in tokenized_docs for word in doc)
)

print("\nVOCABULARY:")
print(vocabulary)

print("\nVOCABULARY SIZE:", len(vocabulary))
```

```python
# -----------------------------------
# STEP 4: TERM FREQUENCY
# -----------------------------------

doc_term_frequencies = [
    Counter(doc)
    for doc in tokenized_docs
]

print("\nD1 TERM FREQUENCIES:")
print(doc_term_frequencies[0])
```

```python
doc_term_frequencies
```

```python
# -----------------------------------
# STEP 5: DOCUMENT FREQUENCY
# -----------------------------------

document_frequency = Counter()

for doc in tokenized_docs:
    document_frequency.update(set(doc))

print("\nDOCUMENT FREQUENCY:")

for word in vocabulary:
    print(word, ":", document_frequency[word])
```

```python
# -----------------------------------
# STEP 6: DOCUMENT LENGTH
# -----------------------------------

document_lengths = [
    len(doc)
    for doc in tokenized_docs
]

average_document_length = (
    sum(document_lengths) / len(document_lengths)
)

print("\nDOCUMENT LENGTHS:")
print(document_lengths)

print("\nAVERAGE DOCUMENT LENGTH:")
print(average_document_length)
```

```python
# -----------------------------------
# STEP 7: BUILD BM25 INDEX
# -----------------------------------

bm25 = BM25Okapi(
    tokenized_docs,
    k1=1.5,
    b=0.75
)

print("\nBM25 INDEX CREATED")
```

```python
# -----------------------------------
# STEP 8: INSPECT IDF
# -----------------------------------

print("\nBM25 IDF VALUES:")

for word in vocabulary:
    print(word, ":", bm25.idf[word])
```

```python
# -----------------------------------
# STEP 9: NEW QUERY
# -----------------------------------

query = "machine learning models"

query_tokens = query.lower().split()

print("\nQUERY TOKENS:")
print(query_tokens)
```

```python
# -----------------------------------
# STEP 10: CALCULATE BM25 SCORES
# -----------------------------------

scores = bm25.get_scores(query_tokens)

print("\nBM25 SCORES:")

for i, score in enumerate(scores):
    print(f"D{i+1}: {score:.4f}")
```

```python
# -----------------------------------
# STEP 11: RANK DOCUMENTS
# -----------------------------------

ranked_results = sorted(
    enumerate(scores),
    key=lambda x: x[1],
    reverse=True
)

print("\nRANKED SEARCH RESULTS:")

for index, score in ranked_results:
    print(
        f"Score: {score:.4f} | "
        f"Document: D{index+1} | "
        f"Text: {documents[index]}"
    )
```

```python
# -----------------------------------
# STEP 12: TOP 3 DOCUMENTS
# -----------------------------------

print("\nTOP 3 DOCUMENTS:")

for index, score in ranked_results[:3]:
    print(documents[index])
    print("Score:", round(score, 4))
    print("-" * 50)
```
